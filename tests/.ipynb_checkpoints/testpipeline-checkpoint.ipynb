```python
sample = dataset[0]
# batch_inputs expects a dict with keys 'img' and 'points'
batch_inputs = {
    'img': sample['inputs']['img'].unsqueeze(0),  # Add batch dim: [1, C, H, W]
    'points': [sample['inputs']['points']]        # Keep points in a list for batch
}

# batch_data_samples is a list of Det3DDataSample
batch_data_samples = [sample['data_samples']]


pred = model(batch_inputs, batch_data_samples)
```

```python
import os

# Define the new directory path
new_dir = "/u/student/2023/cs21m23p100001/mmdetection3d"

# Change the current working directory
os.chdir(new_dir)
from mmdet3d.registry import DATASETS
from mmengine.config import Config
import mmcv
import numpy as np
import matplotlib.pyplot as plt

# Load config
cfg = Config.fromfile('projects/MyFusion/configs/my_fusion_transformer_simple.py')

# Build dataset
dataset = DATASETS.build(cfg.train_dataloader.dataset)

# Load a sample
sample = dataset[0]  # Load first sample
print(sample.keys())
```

```python
print(sample['inputs'])
```

```python
# Extract image and point cloud
image = mmcv.imread(sample['img_path'])
lidar_points = sample['points'].numpy()  # Convert to NumPy for visualization

# Plot Image
plt.figure(figsize=(10, 5))
plt.imshow(image)
plt.title("Loaded Image")
plt.show()

# Plot Point Cloud (Top-down view)
plt.figure(figsize=(10, 5))
plt.scatter(lidar_points[:, 0], lidar_points[:, 1], s=0.5, c=lidar_points[:, 2], cmap='jet')
plt.title("Point Cloud Top-down View")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(label="Z Height")
plt.show()
```
